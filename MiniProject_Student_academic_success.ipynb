# **IQR treatment**

```python
for col in numerical_cols:
  Q1 = df[col].quantile(0.25)
  Q3 = df[col].quantile(0.75)
  IQR = Q3 - Q1

  lower_bound = Q1 - (IQR * 1.5)
  upper_bound = Q3 + (IQR * 1.5)

  lower_outliers = (df[col] < lower_bound).sum()
  upper_outliers = (df[col] > upper_bound).sum()


  print(f"Feature '{col}': Capping {lower_outliers} low & {upper_outliers} high value exceptions.")
  print(f"Safe Bound Window: [{lower_bound:.2f} to {upper_bound:.2f}]")

  df[col] = np.clip(df[col], lower_bound, upper_bound)
```

##**Observation:**
         From the above it is clear that there are some outliers and the outliers can't be removed since the datas are important.Hence the outliers are detected using IQR method and they are replaced using capping method

```python
for col in numerical_cols:
    plt.figure(figsize=(8, 5))
    sns.boxplot(
        data=df,
        y=col

    )

    plt.title(f"{col} Distribution Spread vs. Final Target", fontsize=13, weight="bold")
    plt.xlabel("Academic Status Outcome")
    plt.ylabel(col)
    plt.tight_layout()
    plt.show()
```

## **4.4 Skewness Treatment**
### **Before Transformation**

```python
for col in numerical_cols:
  plt.figure(figsize=(8,6))
  sns.histplot(df[col],kde=True,bins=20)
  plt.show()
```

### **Log / Square Root Transformation**


```python
for col in numerical_cols:
  skewness = df[col].skew()
  print(f'The skewness for the {col} is:{skewness}')

  if skewness > 1:

    df[col] = np.log1p(df[col])


  elif 0.5 < skewness <= 1:
    df[col] = np.sqrt(df[col])


  elif skewness < -0.5:

    max_val = df[col].max()
    reflected_data = max_val + 1 - df[col]

    df[col] = np.log1p(reflected_data)
  else:
    pass

```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC

models = {
    "Random Forest": RandomForestClassifier(class_weight='balanced',random_state=42),
    "KNN (K-Nearest Neighbors)": KNeighborsClassifier(n_neighbors=5),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Naive Bayes": GaussianNB(),
    "Support Vector Machine (SVM)": SVC(class_weight='balanced',kernel='rbf', random_state=42)
}

results = {}
for name, model in models.items():
   model.fit(x_train, y_train)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
evaluation_results = []



for name, model in models.items():
    y_train_pred = model.predict(x_train)
    y_test_pred = model.predict(x_test)

    # 2. Calculate requested Performance Metrics
    metrics = {
        "Train Accuracy": accuracy_score(y_train, y_train_pred),
        "Test Accuracy": accuracy_score(y_test, y_test_pred),
        "Train Precision": precision_score(y_train, y_train_pred,average='macro'),
        "Test Precision": precision_score(y_test, y_test_pred,average='macro'),
        "Train Recall": recall_score(y_train, y_train_pred,average='macro'),
        "Test Recall": recall_score(y_test, y_test_pred,average='macro'),
        "Train F1-Score": f1_score(y_train, y_train_pred,average='macro'),
        "Test F1-Score": f1_score(y_test, y_test_pred,average='macro'),
    }
    evaluation_results.append(metrics)

    # 3. Generate Confusion Matrices
    cm_train = confusion_matrix(y_train, y_train_pred)
    cm_test = confusion_matrix(y_test, y_test_pred)

    # Print individual reports to console
    print(f"  {name} Metrics:")
    print(f"   [Train vs Test Accuracy]: {metrics['Train Accuracy']:.4f} vs {metrics['Test Accuracy']:.4f}")
    print(f"   [Train vs Test F1-Score]: {metrics['Train F1-Score']:.4f} vs {metrics['Test F1-Score']:.4f}")
    print(f"   Test Confusion Matrix:\n  {cm_test}\n")

```

• Severe Overfitting: Tree-based models (Random Forest & Decision Tree) crush the training data (~98% accuracy) but collapse on the test set, highlighting a critical need for hyperparameter regularization (e.g., limiting max_depth).

• The "Enrolled" Blind Spot: Every single model profoundly struggles to predict the middle "Enrolled" tier, with SVM completely failing to capture it (predicting only 1 true positive) while biasing heavily toward the majority "Graduate" class.

```python
from sklearn.utils.class_weight import compute_sample_weight
from xgboost import XGBClassifier

# 1. Calculate individual weights for every single row in your training data
# This forces XGBoost to treat Class 1 as highly important
train_sample_weights = compute_sample_weight(class_weight='balanced', y=y_train)

# 2. Initialize XGBoost for Multi-Class Classification
xgb_model = XGBClassifier(
    objective='multi:softprob',  # Required for multi-class probability outputs
    num_class=3,                 # You have 3 distinct classes (0, 1, 2)
    max_depth=5,                 # Kept low to aggressively fight overfitting
    learning_rate=0.05,          # Slows down learning so it generalizes better
    n_estimators=200,            # Total number of sequential trees
    gamma=1,                     # Minimum loss reduction required to make a split
    random_state=42
)

# 3. Fit the model using the calculated sample weights
xgb_model.fit(x_train, y_train, sample_weight=train_sample_weights)

# 4. Generate predictions
y_train_pred = xgb_model.predict(x_train)
y_test_pred = xgb_model.predict(x_test)

# 5. Calculate Metrics
Train_Accuracy = accuracy_score(y_train, y_train_pred)
Test_Accuracy = accuracy_score(y_test, y_test_pred)

Train_F1_Score = f1_score(y_train, y_train_pred, average='macro')
Test_F1_Score = f1_score(y_test, y_test_pred, average='macro')

cm_test = confusion_matrix(y_test, y_test_pred)

# 6. Print Results
print("XGBoost Classifier Metrics:")
print(f"   [Train vs Test Accuracy]: {Train_Accuracy:.4f} vs {Test_Accuracy:.4f}")
print(f"   [Train vs Test F1-Score]: {Train_F1_Score:.4f} vs {Test_F1_Score:.4f}")
print(f"   Test Confusion Matrix:\n{cm_test}\n")
```

• Top Performer & Best Generalizer: XGBoost achieves the highest Test F1-Score (0.6990) and successfully controls the severe overfitting seen in Random Forest, making it your most reliable model.

• Conquered the Blind Spot: It effectively cracks the ambiguous "Enrolled" tier—correctly identifying 107 students where Random Forest only caught 61 and SVM completely failed.

• The Main Flaw: Its biggest remaining weakness is a tendency to play it safe, misclassifying 79 true Graduates as simply Enrolled.

**1. High-Variance Overfitting (The Tree-Based Trap)**

The standard non-parametric tree models severely memorized the training data noise but collapsed when exposed to unseen student records.

• Decision Tree: Train Acc: 0.9847 vs. Test Acc: 0.6667 (Gap: 31.80%)

• Random Forest (Initial): Train Acc: 0.9847 vs. Test Acc: 0.7638 (Gap: 22.09%)

• Diagnostic: Clear High Variance. The algorithms created highly complex, deep decision boundaries that perfectly separated the training set. However, because student behavior features (like changing grades or fluctuating credit counts) have high intrinsic randomness, these overly complex rules failed to generalize.

**2. Underfitting / High-Bias Classifiers**

These models generalized evenly between datasets but lacked the structural complexity required to accurately separate the complex three-class boundaries.
• Support Vector Machine (SVM): Train F1: 0.5211 vs. Test F1: 0.5234 (Gap: -0.23%)

• Naive Bayes: Train Acc: 0.7290 vs. Test Acc: 0.7266 (Gap: 0.24%)

• Diagnostic: High Bias. While they show near-perfect generalization (zero overfitting gap), their overall predictive power is severely capped. For example, the SVM oversimplified the problem entirely by ignoring the middle "Enrolled" tier and defaulting almost exclusively to predicting the majority "Graduate" class.

**3. Balanced Generalization (The Optimal Zone)**

These algorithms successfully extracted the predictive interactions between academic performance and financial stability features without falling into the memorization trap.

• Logistic Regression: Train Acc: 0.7285 vs. Test Acc: 0.7186 (Gap: 0.99%)

• KNN: Train Acc: 0.7974 vs. Test Acc: 0.7311 (Gap: 6.63%)

• Diagnostic: Optimal Trade-off. XGBoost stands out as the most robust model in this category.
